```python
from keras.models import Sequential
from keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import matplotlib.pyplot as plt
train_dir = "/content/drive/MyDrive/Ảnh F8"
img_width, img_height = 200,200
batch_size = 32

train_datagen = ImageDataGenerator(
    rescale=1.0/255,
    rotation_range=30,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    fill_mode="nearest"
)
train_generator = train_datagen.flow_from_directory(
    train_dir,
    target_size=(img_width, img_height),
    batch_size=batch_size,
    class_mode="categorical"
)
```

```python
model = Sequential([
    Conv2D(32, (3,3), activation="relu", input_shape=(img_width, img_height, 3)),
    MaxPooling2D(2,2),
    Conv2D(64, (3,3), activation="relu"),
    MaxPooling2D(2,2),
    Conv2D(128, (3,3), activation="relu"),
    MaxPooling2D(2,2),
    Flatten(),
    Dense(128, activation="relu"),
    Dropout(0.5),
    Dense(6, activation="softmax")
])
```

```python
model.compile(optimizer="adam",
              loss="categorical_crossentropy",
              metrics=["accuracy"])
model.summary()
epochs = 50
history = model.fit(train_generator, epochs=epochs)
plt.plot(history.history['accuracy'], label="Kết quả huấn luyện")
if 'val_accuracy' in history.history:
    plt.plot(history.history['val_accuracy'], label="Độ chính xác xác thực")

plt.xlabel("Số lần học")
plt.ylabel("Độ chính xác")
plt.legend()
plt.show()
```

```python
import pickle
import os

save_dir = '/content/drive/MyDrive/Ảnh F8'
os.makedirs(save_dir, exist_ok=True)

model_path = os.path.join(save_dir, 'my_face_model.h5')
label_path = os.path.join(save_dir, 'class_labels.pkl')


model.save(model_path)

with open(label_path, 'wb') as f:
    pickle.dump(train_generator.class_indices, f)

print(f"Đã lưu mô hình thành công vào: {model_path}")
print(f"Đã lưu nhãn thành công vào: {label_path}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pickle
from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing.image import load_img, img_to_array

MODEL_PATH = '/content/drive/MyDrive/Ảnh F8/my_face_model.h5'
LABEL_PATH = '/content/drive/MyDrive/Ảnh F8/class_labels.pkl'

IMAGE_PATH = "/content/2.QUE PHUONG 2718596 (1).png"

IMG_SIZE = (200, 200)

model = load_model(MODEL_PATH)
with open(LABEL_PATH, 'rb') as f:
    class_indices = pickle.load(f)
class_labels = {v: k for k, v in class_indices.items()}

img = load_img(IMAGE_PATH, target_size=IMG_SIZE)
plt.imshow(img)
plt.axis('off')
plt.show()


img_array = img_to_array(img)
img_array = img_array / 255.0
img_array = np.expand_dims(img_array, axis=0)

predictions = model.predict(img_array)
prediction_idx = np.argmax(predictions)
person_name = class_labels[prediction_idx]


confidence = predictions[0][prediction_idx] * 100

print(f"Kết quả dự đoán: {person_name} ({confidence:.2f}%)")
```
